# Binary Classification using Deep Neural Network

## IMDB Movie Review Sentiment Analysis

This notebook implements binary classification using a Deep Neural Network (DNN) on the IMDB Movie Reviews dataset.

Goal:
Classify movie reviews into:
- Positive Review
- Negative Review


## Dataset Information

Dataset:
IMDB Movie Reviews Dataset

Contains:
- 50,000 movie reviews
- Binary sentiment labels
- Positive / Negative classification

Dataset is loaded directly from TensorFlow/Keras datasets.

```python
# Import Libraries

import numpy as np
import matplotlib.pyplot as plt

import tensorflow as tf

from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Embedding
from tensorflow.keras.layers import GlobalAveragePooling1D
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

## Load IMDB Dataset

```python
# Load IMDB dataset

vocab_size = 10000

(X_train, y_train), (X_test, y_test) = imdb.load_data(
    num_words=vocab_size
)

print("Training Samples:", len(X_train))
print("Testing Samples:", len(X_test))
```

## Padding Sequences

```python
# Pad sequences to equal length

max_length = 200

X_train = pad_sequences(
    X_train,
    maxlen=max_length
)

X_test = pad_sequences(
    X_test,
    maxlen=max_length
)

print("Training Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)
```

## Build Deep Neural Network

```python
# Build DNN model

model = Sequential([

    Embedding(vocab_size, 32, input_length=max_length),

    GlobalAveragePooling1D(),

    Dense(64, activation='relu'),

    Dense(32, activation='relu'),

    Dense(1, activation='sigmoid')
])

# Compile model

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Model summary

model.summary()
```

## Train Model

```python
# Train model

history = model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=512,
    validation_split=0.2,
    verbose=1
)
```

## Evaluate Model

```python
# Evaluate model

loss, accuracy = model.evaluate(
    X_test,
    y_test
)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```

## Predictions

```python
# Predict sentiment probabilities

predictions = model.predict(X_test)

# Convert probabilities into labels

predicted_labels = (predictions > 0.5).astype(int)

print("First 10 Predictions:")

for i in range(10):

    sentiment = "Positive" if predicted_labels[i][0] == 1 else "Negative"

    print(f"Review {i+1}: {sentiment}")
```

## Visualization

```python
# Plot training and validation accuracy

plt.figure(figsize=(8,6))

plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])

plt.title("Model Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.legend(['Train', 'Validation'])

plt.show()
```

## Loss Visualization

```python
# Plot training and validation loss

plt.figure(figsize=(8,6))

plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])

plt.title("Model Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.legend(['Train', 'Validation'])

plt.show()
```
