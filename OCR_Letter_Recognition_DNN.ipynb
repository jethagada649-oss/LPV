# Multiclass Classification using Deep Neural Network

## OCR Letter Recognition Dataset

This notebook implements multiclass classification using a Deep Neural Network (DNN) on the OCR Letter Recognition dataset.

Dataset Source:
https://archive.ics.uci.edu/ml/datasets/letter+recognition


## Dataset Information

The dataset contains:
- 20,000 samples
- 16 numerical features
- 26 output classes (A-Z)

Goal:
Predict the alphabet letter using feature values.

```python
# Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.utils import to_categorical
```

## Load Dataset

Download dataset:
https://archive.ics.uci.edu/ml/machine-learning-databases/letter-recognition/letter-recognition.data

Save as:
`letter-recognition.csv`


```python
# Column Names

columns = [
    'letter',
    'x-box',
    'y-box',
    'width',
    'height',
    'onpix',
    'x-bar',
    'y-bar',
    'x2bar',
    'y2bar',
    'xybar',
    'x2ybr',
    'xy2br',
    'x-ege',
    'xegvy',
    'y-ege',
    'yegvx'
]

# Load dataset

df = pd.read_csv(
    "letter-recognition.csv",
    names=columns
)

# Display first 5 rows

df.head()
```

## Dataset Information

```python
# Dataset shape

print("Dataset Shape:", df.shape)

# Dataset info

df.info()
```

## Feature Selection

```python
# Features and Labels

X = df.drop("letter", axis=1)

y = df["letter"]

print("Feature Shape:", X.shape)
print("Label Shape:", y.shape)
```

## Label Encoding

```python
# Encode labels

encoder = LabelEncoder()

y_encoded = encoder.fit_transform(y)

# One-hot encoding

y_categorical = to_categorical(y_encoded)

print("Encoded Labels Shape:", y_categorical.shape)
```

## Train-Test Split

```python
# Split dataset

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_categorical,
    test_size=0.2,
    random_state=42
)

print("Training Samples:", X_train.shape[0])
print("Testing Samples:", X_test.shape[0])
```

## Feature Scaling

```python
# Standardization

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

## Build Deep Neural Network

```python
# Build DNN model

model = Sequential([
    Dense(128, activation='relu', input_shape=(16,)),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(26, activation='softmax')
])

# Compile model

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Model summary

model.summary()
```

## Train Model

```python
# Train model

history = model.fit(
    X_train,
    y_train,
    epochs=50,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)
```

## Evaluate Model

```python
# Evaluate model

loss, accuracy = model.evaluate(X_test, y_test)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```

## Predictions

```python
# Predict classes

y_pred = model.predict(X_test)

# Convert probabilities to class labels

y_pred_classes = np.argmax(y_pred, axis=1)
y_test_classes = np.argmax(y_test, axis=1)

# Accuracy

acc = accuracy_score(y_test_classes, y_pred_classes)

print("Accuracy Score:", acc)
```

## Classification Report

```python
# Classification report

print(
    classification_report(
        y_test_classes,
        y_pred_classes
    )
)
```

## Confusion Matrix

```python
# Confusion matrix

cm = confusion_matrix(
    y_test_classes,
    y_pred_classes
)

print(cm)
```

## Visualization

```python
# Plot training accuracy

plt.figure(figsize=(8,6))

plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])

plt.title("Model Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.legend(['Train', 'Validation'])

plt.show()
```
